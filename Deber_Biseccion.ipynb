```python
import math

# Implementation del Algoritmo 2.1
def biseccion_algoritmo_2_1(f, a, b, TOL, N0=100):
    i = 1
    FA = f(a)
    
    while i <= N0:
        p = a + (b - a) / 2.0
        FP = f(p)
        
        # Paso 4: parada por tolerancia o raíz exacta
        if FP == 0 or (b - a) / 2.0 < TOL:
            return p, i
        
        i += 1
        
        # Paso 6: actualización de intervalo
        if FA * FP > 0:
            a = p
            FA = FP
        else:
            b = p
            
    return None, N0

#Sección 1
f1 = lambda x: x**3 - x - 1
p1, i1 = biseccion_algoritmo_2_1(f1, 1.0, 2.0, 1e-4)
print(f"Sección 1.3 -> Solución: {p1:.6f} | Iteraciones: {i1}")

#Sección 2
f2 = lambda x: x**3 - 7*x**2 + 14*x - 6
p2_1, i2_1 = biseccion_algoritmo_2_1(f2, 0.0, 1.0, 1e-2)
p2_2, i2_2 = biseccion_algoritmo_2_1(f2, 1.0, 3.2, 1e-2)
p2_3, i2_3 = biseccion_algoritmo_2_1(f2, 3.2, 4.0, 1e-2)

print(f"Sección 2.1 [0, 1]    -> Solución: {p2_1:.6f} | Iteraciones: {i2_1}")
print(f"Sección 2.1 [1, 3.2]  -> Solución: {p2_2:.6f} | Iteraciones: {i2_2}")
print(f"Sección 2.1 [3.2, 4]  -> Solución: {p2_3:.6f} | Iteraciones: {i2_3}")

# Intersecciones
f3_a = lambda x: x - 2*math.sin(x)
p3_a, _ = biseccion_algoritmo_2_1(f3_a, 1.0, math.pi, 1e-5)

f3_b = lambda x: x - math.tan(x)
p3_b, _ = biseccion_algoritmo_2_1(f3_b, 4.0, 4.6, 1e-5)

f3_c = lambda x: x**2 - 1 - math.exp(1 - x**2)
p3_c, _ = biseccion_algoritmo_2_1(f3_c, -2.0, 0.0, 1e-3)

print(f"Sección 2.3 (x = 2*sin x)  -> Solución: {p3_a:.6f}")
print(f"Sección 2.3 (x = tan x)    -> Solución: {p3_b:.6f}")
print(f"Sección 2.3 (x^2-1 = e...) -> Solución: {p3_c:.6f}")

#Sección 3

f_mult = lambda x: (x + 3) * ((x + 1)**2) * x * ((x - 1)**3) * (x - 3)
p4_1, _ = biseccion_algoritmo_2_1(f_mult, -3.5, 2.5, 1e-5)
p4_2, _ = biseccion_algoritmo_2_1(f_mult, -1.5, 0.6, 1e-5)
p4_3, _ = biseccion_algoritmo_2_1(f_mult, 0.4, 2.2, 1e-5)
p4_4, _ = biseccion_algoritmo_2_1(f_mult, 2.0, 3.7, 1e-5)

print(f"Sección 3.1 [-3.5, 2.5] -> Solución: {p4_1:.6f}")
print(f"Sección 3.1 [-1.5, 0.6] -> Solución: {p4_2:.6f}")
print(f"Sección 3.1 [0.4, 2.2]  -> Solución: {p4_3:.6f}")
print(f"Sección 3.1 [2.0, 3.7]  -> Solución: {p4_4:.6f}")

#Sección 4
L, r, V = 10.0, 1.0, 12.4
f_tanque = lambda h: L * (0.5 * math.pi * (r**2) - (r**2) * math.asin(h / r) - h * math.sqrt(r**2 - h**2)) - V
h_sol, _ = biseccion_algoritmo_2_1(f_tanque, 0.001, r - 0.001, 0.01)

# Caída libre
s0, m, k, g = 300.0, 0.25, 0.1, 9.81
s_t = lambda t: s0 - (m * g / k) * t + ((m**2 * g) / (k**2)) * (1 - math.exp(-k * t / m))
t_sol, _ = biseccion_algoritmo_2_1(s_t, 0.0, 100.0, 0.01)

print(f"Sección 4.1 (Abrevadero h)  -> {h_sol:.2f} cm")
print(f"Sección 4.1 (Profundidad)   -> {r - h_sol:.2f} cm")
print(f"Sección 4.2 (Caída libre t) -> {t_sol:.2f} s")
```
