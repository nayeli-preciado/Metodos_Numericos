```python
from scipy.optimize import newton

def funcion(x): 
 return x**3 + 4*x**2 -10


def derivada(x):
 return 3*x**2 + 8*x

newton(func=funcion , x0=1.5, fprime=derivada, args=(), tol=1.48E-08, maxiter=50)

```
